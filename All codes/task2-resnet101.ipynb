```python

import os
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
import random
import tensorflow as tf
from tensorflow.keras.models import Model, load_model
from sklearn.model_selection import train_test_split
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Dropout
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from tensorflow.keras.applications import ResNet101
from tensorflow.keras.optimizers import Adamax
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc
from sklearn.preprocessing import LabelBinarizer
from itertools import cycle

# Set a seed for reproducibility
random.seed(42)
np.random.seed(42)
tf.random.set_seed(42)

# -------------------------------------------------------------------------
# 1. DATA LOADING AND EXPLORATION
# -------------------------------------------------------------------------
dataset_path = '/kaggle/input/multi-cancer/Multi Cancer/Multi Cancer/Cervical Cancer'
cervix_categories = ['cervix_dyk', 'cervix_koc', 'cervix_mep', 'cervix_pab', 'cervix_sfi']

filepaths = []
labels = []

for category in cervix_categories:
    folder_path = os.path.join(dataset_path, category)
    if os.path.isdir(folder_path):
        for fname in os.listdir(folder_path):
            filepaths.append(os.path.join(folder_path, fname))
            labels.append(category)

# Create DataFrame
df = pd.DataFrame({'filepath': filepaths, 'labels': labels})

print("DataFrame Head:")
print(df.head())
print("\nDataset Info:")
df.info()
print("\nValue Counts:")
print(df['labels'].value_counts())


# -------------------------------------------------------------------------
# 2. DATA VISUALIZATION
# -------------------------------------------------------------------------
# Set the style and color palette
sns.set(style="whitegrid")
counts = df['labels'].value_counts()
palette = sns.color_palette("coolwarm", n_colors=len(counts))

# Create the bar plot
plt.figure(figsize=(12, 8))
bar_plot = sns.barplot(x=counts.index, y=counts.values, palette=palette)
plt.title('The Count of Each Category', fontsize=20, fontweight='bold')
plt.xlabel('Labels', fontsize=16)
plt.ylabel('Count', fontsize=16)

# Add value annotations
for p in bar_plot.patches:
    bar_plot.annotate(f'{int(p.get_height())}',
                      (p.get_x() + p.get_width() / 2., p.get_height()),
                      ha='center', va='bottom', fontsize=12, color='black',
                      xytext=(0, 5), textcoords='offset points')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# -------------------------------------------------------------------------
# 3. DATA SPLITTING AND GENERATORS
# -------------------------------------------------------------------------
# Stratified split into train, validation, and test sets
strat = df['labels']
train_df, dummy_df = train_test_split(df, test_size=0.3, random_state=42, stratify=strat)
valid_df, test_df = train_test_split(dummy_df, test_size=0.5, random_state=42, stratify=dummy_df['labels'])

print(f"Train set shape: {train_df.shape}")
print(f"Validation set shape: {valid_df.shape}")
print(f"Test set shape: {test_df.shape}")

# Image Data Generators
batch_size = 64
img_size = (224, 224)

# No augmentation for simplicity, can be added to tr_gen
tr_gen = ImageDataGenerator()
ts_gen = ImageDataGenerator()

train_gen = tr_gen.flow_from_dataframe(
    train_df, x_col='filepath', y_col='labels', target_size=img_size,
    class_mode='categorical', color_mode='rgb', shuffle=True, batch_size=batch_size
)

valid_gen = ts_gen.flow_from_dataframe(
    valid_df, x_col='filepath', y_col='labels', target_size=img_size,
    class_mode='categorical', color_mode='rgb', shuffle=True, batch_size=batch_size
)

test_gen = ts_gen.flow_from_dataframe(
    test_df, x_col='filepath', y_col='labels', target_size=img_size,
    class_mode='categorical', color_mode='rgb', shuffle=False, batch_size=batch_size
)


# -------------------------------------------------------------------------
# 4. MODEL BUILDING (ResNet101)
# -------------------------------------------------------------------------
# Load base model with pre-trained ImageNet weights
base_model = ResNet101(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
base_model.trainable = False  # Freeze the base model layers

# Create a new sequential model
model3 = Sequential([
    base_model,
    Flatten(),
    Dense(256, activation='relu'),
    Dropout(0.25),
    Dense(64, activation='relu'),
    Dropout(0.25),
    Dense(len(cervix_categories), activation='softmax')  # Output layer for 5 classes
])

# Compile the model
optimizer = Adamax(learning_rate=0.002)
model3.compile(optimizer=optimizer, loss='categorical_crossentropy', metrics=['accuracy'])
model3.summary()


# -------------------------------------------------------------------------
# 5. MODEL TRAINING
# -------------------------------------------------------------------------
# Define callbacks
early_stopping = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
model_checkpoint = ModelCheckpoint(
    'best_model.keras', monitor='val_loss', save_best_only=True, verbose=1
)

# Fit the model
history = model3.fit(
    train_gen,
    epochs=50,
    validation_data=valid_gen,
    callbacks=[early_stopping, model_checkpoint]
)


# =========================================================================
# 6. COMPREHENSIVE MODEL EVALUATION (UPDATED SECTION)
# =========================================================================

print("\n--- Starting Final Model Evaluation ---")

# Load the best performing model from the checkpoint
best_model = load_model('best_model.keras')

# --- Test Inference Time ---
print("Calculating inference time...")
start_time = time.time()
y_pred_prob = best_model.predict(test_gen, verbose=1)
end_time = time.time()
test_time = end_time - start_time
print(f"\nTest Inference Time: {test_time:.2f} seconds")

# Get predicted class labels and true labels
y_pred_classes = np.argmax(y_pred_prob, axis=1)
y_true = test_gen.classes
class_labels = list(test_gen.class_indices.keys())
n_classes = len(class_labels)

# --- Classification Report ---
print("\nClassification Report:")
print(classification_report(y_true, y_pred_classes, target_names=class_labels))

# --- Per-Class Accuracy ---
cm = confusion_matrix(y_true, y_pred_classes)
print("Per-Class Accuracy:")
for i, label in enumerate(class_labels):
    class_accuracy = cm[i, i] / cm[i, :].sum()
    print(f"Class '{label}' Accuracy: {class_accuracy:.3f}")

# --- ROC Curve and AUC for Multi-Class ---
# Binarize the labels
y_true_bin = LabelBinarizer().fit_transform(y_true)

# Dictionaries to store ROC curve data
fpr = dict()
tpr = dict()
roc_auc = dict()

for i in range(n_classes):
    fpr[i], tpr[i], _ = roc_curve(y_true_bin[:, i], y_pred_prob[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])

# Plot all ROC curves
plt.figure(figsize=(12, 8))
colors = cycle(['aqua', 'darkorange', 'cornflowerblue', 'green', 'red'])
for i, color in zip(range(n_classes), colors):
    plt.plot(fpr[i], tpr[i], color=color, lw=2,
             label=f'ROC curve of class {class_labels[i]} (area = {roc_auc[i]:0.2f})')

plt.plot([0, 1], [0, 1], 'k--', lw=2)
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate', fontsize=14)
plt.ylabel('True Positive Rate', fontsize=14)
plt.title('Receiver Operating Characteristic (ROC) for Multi-Class', fontsize=16)
plt.legend(loc="lower right")
plt.show()


# --- Displaying Sample Predictions ---
def display_prediction_samples(generator, model, num_samples=5):
    # Get a single batch from the generator
    x_batch, y_batch_true = next(generator)
    
    # Make predictions on that batch
    y_batch_pred_prob = model.predict(x_batch, verbose=0)
    y_batch_pred = np.argmax(y_batch_pred_prob, axis=1)
    y_batch_true = np.argmax(y_batch_true, axis=1)

    plt.figure(figsize=(20, 8))
    for i in range(min(num_samples, len(x_batch))):
        plt.subplot(1, num_samples, i + 1)
        plt.imshow((x_batch[i] * 255).astype(np.uint8) if x_batch.max() <= 1.0 else x_batch[i].astype(np.uint8))
        
        actual_label = class_labels[y_batch_true[i]]
        predicted_label = class_labels[y_batch_pred[i]]
        
        title_color = 'green' if actual_label == predicted_label else 'red'
        plt.title(f"Actual: {actual_label}\nPredicted: {predicted_label}", color=title_color)
        plt.axis('off')
    plt.tight_layout()
    plt.show()

print("\n--- Displaying a few sample images with their predicted labels ---")
# Reset generator before using it to ensure it starts from the beginning
test_gen.reset()
display_prediction_samples(test_gen, best_model)
```

```python
# -------------------------------------------------------------------------
# CONFUSION MATRIX VISUALIZATION
# -------------------------------------------------------------------------
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=class_labels, yticklabels=class_labels)
plt.ylabel('Actual Label', fontsize=12)
plt.xlabel('Predicted Label', fontsize=12)
plt.title('Confusion Matrix', fontsize=16, fontweight='bold')
plt.show()
```
